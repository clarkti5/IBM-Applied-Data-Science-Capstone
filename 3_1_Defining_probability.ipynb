## Probability

We define probability in terms of a **random process** producing an **outcome**. Common examples include

* Rolling one or more die/dice.
* Flipping a coin.
* Randomly selecting a card from a well-shuffled deck.

You may notice a lot of the commonnly present examples involve gambling—this is not a coincidence! The mathematical study of probability was historically motivated by studying various games of chance.

Note that a set or collection of outcomes is called an **event**, while the set of all possible outcomes is called the **sample space**. Observing a single outcome from a random process is called a **trial**.

### Definition of Probability
The **probability** of a given outcome or event arising  from a random process is the proportion of times the
outcome or event would occur if we were able to observe the random process an infinite number of times.

The probability of an event always takes a value of at
least `0%` and no more than `100%`.

### Question

If you flip a (fair) coin, what do you think the probability of getting `heads` should be?

We can do a simulation in Python of flipping a coin to see how the proportion of `heads` changes as the number of coin flips increases.

```python
import warnings
warnings.filterwarnings("ignore")

import math
import pandas as pd
import numpy as np
from plotnine import *
```

We set the outcomes of our coin flip as `heads` and `tails`.

```python
coin_flip_outcomes = np.array(['heads', 'tails'])
```

```python
np.random.choice(coin_flip_outcomes)
```

We can then use `random.choice` to simulate making 100 coin flips and counting the proportion of `heads` observed.

```python
proportions = []

for num_flips in range(100):
  flips = np.random.choice(coin_flip_outcomes, size = num_flips, replace = True, p = [0.50, 0.50])
  heads_count = (flips == 'heads').sum()
  heads_proportion = heads_count / num_flips
  proportions.append(heads_proportion)

(
    ggplot(pd.DataFrame({'proportion': proportions}), aes(x = pd.DataFrame({'proportion': proportions}).index, y = 'proportion')) +
    geom_line() +
    geom_hline(yintercept = 0.5, linetype = 'dashed')
)
```

Let's do it again with 1000 coin flips!

```python
proportions = []

for num_flips in range(1000):
  flips = np.random.choice(coin_flip_outcomes, size = num_flips, replace = True, p = [0.50, 0.50])
  heads_count = (flips == 'heads').sum()
  heads_proportion = heads_count / num_flips
  proportions.append(heads_proportion)

(
    ggplot(pd.DataFrame({'proportion': proportions}), aes(x = pd.DataFrame({'proportion': proportions}).index, y = 'proportion')) +
    geom_line() +
    geom_hline(yintercept = 0.5, linetype = 'dashed')
)
```

### The law of large numbers

The proportion of occurence of a particular outcome
after $n$ trials is denoted $ \hat{p_n}$ (sometimes
called the **relative frequency**).

As $n$ increases toward infinity, $ \hat{p_n} $
converges to the probability of the outcome, $p$. In
other words
$$ \hat{p_n} \to p \text{ as } n \to \infty.$$

We can summarize our findings with something called the **Addition Rule for Disjoint Events**.

### Addition rule for disjoint events

Let $A$ and $B$ be disjoint. Then
$$ P (A \text{ or } B) = P(A) + P(B).$$

A similar statement holds in the case of more than 2
disjoint outcomes or events.

### General addition rule

If $A$ and $B$ are *any* two events, then

$$ P(A \text{ or } B) = P(A) + P(B) - P(A \text{ and } B).$$

Venn diagrams are extremely useful and motivate the following terminology.

* The event "$A \text{ or } B$" is sometimes called the **union** of events $A$ and $B$, denoted $A \cup B$.

  * Note: "or" is used inclusively in mathematics, i.e., you should think of "or" as "and/or".
  * So, $A \cup B$ is a shorthand for saying "$A$ or $B$, or both."

* The event "$A \text{ and } B$" is sometimes called the **intersection** of events $A$ and $B$, denoted $A \cap B$.

## Probability distributions

A **probability distribution** is an exhaustive list of outcomes and their probabilities for a particular random process.

The probability distribution must list *all* possible outcomes with their corresponding probabilities such that
1. The outcomes listed are all disjoint.
2. Each probability has a value between 0 and 1.
3. The sum of the probabilities is equal to 1.

### Example

Consider the process of rolling two dice and recording the sum of the values facing up. What is the probability distribution?

### Probability of a complement

Given an event $A$, the **complement** of $A$ is denoted $A^c$ and it represents all of the outcomes that are
*not* in $A$.

Since the event "$A \text{ or } A^c$" represents the
entire sample space and $A$ is disjoint from $A^c$ we
find
* $P(A \text{ or } A^c) = 1.$
* $P(A) + P(A^c) = 1.$
* $P(A) = 1 - P(A^c)$ and $P(A^c) = 1 - P(A).$

## Independence

Two processes are called **independent** if knowing the outcome of one process provides you no useful information in determining the outcome of the other process.

### Question

Suppose you flip a coin 3 times and record the side facing up.

* Does the outcome of the first coin flip in any way influence the outcome of the second coin flip? In other words, is each coin flip independent of the others?
* What is the probability of getting `heads` for all 3 coin flips?

The results of our investigation above may not be very intuitive. We can use a Python simulation to see it in action! The code below will simulate flipping a coin 3 times.

```python
three_flips = np.random.choice(coin_flip_outcomes, size = 3, replace = True, p = [0.50, 0.50])
three_flips
```

Now, we can repeat the process above 1000 times and examine the proportion of times that we observe `heads` on all 3 flips.

```python
trial_results = []
all_heads_proportions = []

for num_trials in range(1, 1001):
  trial_results.append(np.random.choice(coin_flip_outcomes, size = 3, replace = True, p = [0.50, 0.50]).tolist())
  all_heads_count = trial_results.count(['heads','heads','heads'])
  all_heads_proportion = all_heads_count / num_trials
  all_heads_proportions.append(all_heads_proportion)

(
    ggplot(pd.DataFrame({'proportion': all_heads_proportions}), aes(x = pd.DataFrame({'proportion': all_heads_proportions}).index, y = 'proportion')) +
    geom_line() +
    geom_hline(yintercept = 0.125, linetype = 'dashed')
)
```

### Multiplication rule for independent processes

If $A$ and $B$ represent events from two independent
processes, then
$$ P(A \text{ and } B) = P(A) \cdot P(B).$$

A similar result holds for more than 2 events from
independent processes.
