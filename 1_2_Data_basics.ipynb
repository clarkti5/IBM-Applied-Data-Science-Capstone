## Observations, variables, and data matrices

The first five rows of this data set are displayed below. Note that each row represents a single loan (sometimes called a **case** or **observational unit**) and each column represents a **variable** related to that loan (sometimes called a **feature**).

```python
loan50.head()
```

This data structure is called a **data matrix**. It looks like a spreadsheet, but is a little more organized.

- Each row has a special **index**.
- Each column has a fixed **data type** --- in a spreadsheet, you can mix and match data types within a column, though it usually isn't considered good practice!


## Variable types

Variables come in different types.

- **Numerical** variables are those on which you can perform arithmetic (addition, subtraction, averages, etc.) and find a reasonable result.
  - Numerical variables can be further classified as **continuous** (able to take on any real number value) or **discrete** (only able to be whole numbers).
- **Categorical** variables, on the other hand, arrange the data into categories, groups, or **levels**. Though the levels may be represented by a number (e.g. 0 or 1), arithmetic does not produce sensible results from categorical variables.

## Relationships between variables

### Questions

- Are the variables `annual_income` and `loan_amount` **related** (or **associated**) to one another?
- What about `emp_length` and `total_credit_utilized`?

```python
(
    ggplot(loan50) +
    aes(x = 'annual_income', y = 'loan_amount') +
    geom_point()
   + geom_smooth(method = 'lm', se = False)
)
```

When we think one variable might be affecting another, we call the variables **associated**. Typically, we'll think of one variable as the **explanatory variable** and the other as the **response variable**. If making a scatterplot, as done above, we plot the explanatory variable on the $x$-axis and the response variable on the $y$-axis.

## Observations and experiments

The manner in which data is a collected will impact the sorts of conclusions we can draw from it. A common saying in data analytics is

> Garbage in, garbage out!

There are two main data collection strategies we'll see in this course.

- **Observational studies** collect data in a way that does not
directly interfere with how the data arise (e.g. through surveys or a review of medical records). Observational studies can give us a glimpse into naturally arising associations, but we *cannot* infer causal relationships from them.
- **Randomized experiments** randomly assign participants into groups (often called `treatment` and `control` groups). By controlling for hidden variables, a well-designed experiment *can* be used to infer causal relationships. We'll study experiments more in Section 1.4.
